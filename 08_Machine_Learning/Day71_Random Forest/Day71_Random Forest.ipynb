# Random Forest
Definition: Random Forest is a supervised Machine Learning algorithm that combines multiple Decision Trees to make a prediction.

For classification: \
Many Trees \
     ↓ \
Voting \
     ↓ \
Final Class

*For example:* \
Tree 1 → Pass \
Tree 2 → Pass \
Tree 3 → Fail \
Tree 4 → Pass \
Tree 5 → Pass

Final → Pass

## Why Not Just Use One Decision Tree?
A single Decision Tree can sometimes become too focused on the training data.

This is called:**Overfitting**

*For example:*

Training Data \
     ↓ \
Decision Tree \
     ↓ \
Very complicated rules \
     ↓ \
Excellent training result \
     ↓ \
Poor new-data result

Random Forest reduces this problem by using many different trees and combining their predictions.

## Decision Tree vs Random Forest

| Decision Tree      | Random Forest                         |
| ------------------ | ------------------------------------- |
| One tree           | Many trees                            |
| Can overfit easily | Usually more resistant to overfitting |
| Simple             | More powerful                         |
| Fast to understand | More complex internally               |
| One prediction     | Many predictions → combined           |

### Why "Forest"?
Think about it: One Tree and many trees. Many trees together form a **Forest**

So, Random + Forest

## Why "Random"?

Random Forest introduces randomness when creating the trees. \
Different trees can learn from different samples and feature subsets. \
This means the trees don't all behave exactly the same way. \
That's useful because if every tree made exactly the same mistakes, voting wouldn't help much.

**Complete Code**

```python
import pandas as pd

from sklearn.ensemble import RandomForestClassifier

data = {
    "Hours": [2,3,4,6,7,8,9,10],
    "Attendance": [60,65,70,80,85,90,92,95],
    "Pass": [0,0,0,1,1,1,1,1]
}

df = pd.DataFrame(data)

X = df[["Hours", "Attendance"]]

y = df["Pass"]

model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

model.fit(X, y)

prediction = model.predict([[7, 88]])

print("Prediction:", prediction)

if prediction[0] == 1:
    print("Student will Pass")
else:
    print("Student will Fail")
```

* n_estimators

*model = RandomForestClassifier( \
    n_estimators=100 \
)*

n_estimators means: How many Decision Trees should the Random Forest create?

For example: n_estimators=10 means approximately: 10 Trees & n_estimators=100 means 100 Trees

Think: n_estimators = number of trees

_What Happens If We Increase It?_
For example: n_estimators=500 

Generally: \
More trees can make predictions more stable. \
But more trees require more computation. \
We don't need to use 1,000 trees for every problem.

A value such as 100 is a reasonable starting point for learning.

* random_state

*random_state=42* It makes the random process repeatable.

Without it: \
Random Forest \
     ↓ \
Random choices \
     ↓ \
Potentially different result

With it: \
random_state=42 \
     ↓ \
Same random process \
     ↓ \
Reproducible result 

The number 42 itself isn't special. \
We could use: *random_state=10* or *random_state=100*

The important thing is that you use the same value when you want reproducible results.


## Checking Accuracy
We can evaluate the model.

```python
from sklearn.metrics import accuracy_score

predictions = model.predict(X)

accuracy = accuracy_score(y, predictions)

print("Accuracy:", accuracy)
```

Important: This is training-data accuracy because we're predicting on X, the same data used to train the model. \
A training accuracy of 1.0 does not automatically mean the model will perform perfectly on new data.

**Better Version (Train/Test Split)**

```python
import pandas as pd

from sklearn.ensemble import RandomForestClassifier

from sklearn.model_selection import train_test_split

from sklearn.metrics import accuracy_score

data = {
    "Hours": [2,3,4,6,7,8,9,10],
    "Attendance": [60,65,70,80,85,90,92,95],
    "Pass": [0,0,0,1,1,1,1,1]
}

df = pd.DataFrame(data)

X = df[["Hours", "Attendance"]]

y = df["Pass"]

X_train, X_test, y_train, y_test = train_test_split(  # Splits Data in Training and Testing Data 
    X,
    y,
    test_size=0.25,
    random_state=42
)

model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

model.fit(X_train, y_train)       # Learn from the training data.

y_pred = model.predict(X_test)   # Make predictions on data the model didn't train on.

accuracy = accuracy_score(y_test, y_pred)   # compares: Actual Test Answers vs Predicted Answers

print("Accuracy:", accuracy)
```

## Feature Importance

One very useful thing about Random Forest is that it can tell us which features were important.

Use:

```python
print(model.feature_importances_)
```

Our features were: \
Hours \
Attendance

So conceptually: \
Hours      → 0.50 \
Attendance → 0.49
    
This suggests Hours contributed more to the model's decision than Attendance for this fitted dataset.

#### Match Features With Importance

```python
for feature, importance in zip(X.columns, model.feature_importances_):
    print(feature, ":", importance)
```

zip() pairs the feature name with its corresponding importance value.

**_Very Important_**

Feature importance does not automatically mean: "Hours causes Marks to increase by 65%." \
It means the feature was relatively useful to the trained Random Forest when making its splits. \
Don't interpret feature importance as causation.

## Mini Project (Loan Approval)

We'll use: \
Income \
Credit Score \
Loan Approved

```python
import pandas as pd

from sklearn.ensemble import RandomForestClassifier

data = {
    "Income": [
        20000, 25000, 30000, 40000,
        50000, 60000, 70000, 90000
    ],

    "Credit": [
        500, 550, 600, 650,
        700, 720, 760, 800
    ],

    "Approved": [
        0, 0, 0, 0,
        1, 1, 1, 1
    ]
}

df = pd.DataFrame(data)

X = df[["Income", "Credit"]]

y = df["Approved"]

model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

model.fit(X, y)

income = float(input("Enter your income: "))

credit = float(input("Enter your credit score: "))

prediction = model.predict([[income, credit]])

if prediction[0] == 1:
    print("Loan Approved")
else:
    print("Loan Rejected")
```

This is only a learning example. \
A real loan approval system would require much more data, careful validation, fairness checks, appropriate features, and proper financial/regulatory considerations.
