# Multiple Linear Regression

## What is Multiple Linear Regression? 
Definition: Multiple Linear Regression is a Machine Learning algorithm that predicts a continuous numerical value using two or more input features.

For Example: \
Suppose teacher wants to predict a student's marks. \
She first uses only: Study Hour \
But then she thinks: Marks dont depend only on study hours.\
They Also depend on: Attendance, Assignements, Study Hours.

Now She uses all three. This is **Multiple Linear Regression**.

## Linear Regression vs Multiple Linear Regression

| Linear Regression | Multiple Linear Regression               |
| ----------------- | ---------------------------------------- |
| 1 Input Feature   | 2 or More Input Features                 |
| Hours → Marks     | Hours + Attendance + Assignments → Marks |

### Real-Life Examples
**Predict House Price** \
Inputs: \
House Size \
Bedrooms \
Bathrooms \
Age of House

Output: \
House Price

**Predict Student Marks** \
Inputs: \
Study Hours \
Attendance \
Assignments 

Output: \
Marks

### Mathematical Equation
Multiple Linear Regression extends the simple equation. \
Instead of: \
y = mx + b

We use: \
y = b + m₁x₁ + m₂x₂ + m₃x₃ + ...

Where: \
y = Predicted value \
b = Intercept \
x₁ = Hours \
x₂ = Attendance \
x₃ = Assignments \
m₁, m₂, m₃ = Slopes (weights)

Just remember: Each input feature contributes to the prediction.

### Our Dataset

```python
import pandas as pd

data = {
    "Hours": [2,4,6,8,10],
    "Attendance": [60,70,80,90,95],
    "Assignments": [3,5,6,8,10],
    "Marks": [40,55,70,85,95]
}

df = pd.DataFrame(data)

print(df)
```

## Step 1 – Import Libraries

```python
import pandas as pd

from sklearn.linear_model import LinearRegression
```

## Step 2 – Create Features (X)

```python
X = df[["Hours", "Attendance", "Assignments"]]
```

#### Why three columns? 
Yesterday: X = df[["Hours"]] \
Only one feature.

Today: X = df[["Hours", "Attendance", "Assignments"]] \
Three features.

## Step 3 – Create Target (y)

```python
y = df["Marks"]
print(y)
```

## Step 4 – Create Model

```python
model = LinearRegression()
```

The algorithm is the same. \
The only difference is that it now learns from multiple columns.

## Step 5 – Train Model

```python
model.fit(X, y)
```

What happens? \
The model learns: \
How Hours affect Marks. \
How Attendance affects Marks. \
How Assignments affect Marks. 

It combines all three to make predictions.

## Step 6 – Predict

Suppose a new student has: \
Hours = 7 \
Attendance = 85 \
Assignments = 7 

Predict:

```python
prediction = model.predict([[7,85,7]])

print(prediction)
```

Meaning: The model predicts the student may score about 78 marks.

### Why [[7,85,7]]?

Remember: \
During training:

```python
X = df[["Hours", "Attendance", "Assignments"]]
```

There were 3 columns. \
So during prediction, you must also provide 3 values. \
Correct: [[7,85,7]] \
Meaning:  

| Hours | Attendance | Assignments |
| ----: | ---------: | ----------: |
|     7 |         85 |           7 |

Always use: [[7,85,7]]

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

data = {
    "Hours": [2,4,6,8,10],
    "Attendance": [60,70,80,90,95],
    "Assignments": [3,5,6,8,10],
    "Marks": [40,55,70,85,95]
}

df = pd.DataFrame(data)

X = df[["Hours", "Attendance", "Assignments"]]
y = df["Marks"]

model = LinearRegression()

model.fit(X, y)

prediction = model.predict([[7,85,7]])

print("Predicted Marks:", prediction[0])
```

## Understanding model.coef_

After training:

print(model.coef_)

Example Output:\
[4.5 0.6 2.1] \

These numbers tell us how much each feature influences the prediction. \
For example:

| Feature     | Coefficient |
| ----------- | ----------: |
| Hours       |         4.5 |
| Attendance  |         0.6 |
| Assignments |         2.1 |

A larger positive coefficient generally means that increasing that feature increases the predicted marks more (assuming other features stay the same).

## Understanding model.intercept_

print(model.intercept_) \
Example: 5.2

This is the starting value (the intercept) in the regression equation.

## Mini Project (Salary Prediction)

Predict salary using: 
* Experience
* Education Years
* Certifications

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

data = {
    "Experience": [1,2,3,4,5],
    "Education": [12,14,16,16,18],
    "Certificates": [0,1,1,2,3],
    "Salary": [25000,35000,45000,60000,80000]
}

df = pd.DataFrame(data)

X = df[["Experience", "Education", "Certificates"]]
y = df["Salary"]

model = LinearRegression()

model.fit(X, y)

salary = model.predict([[6,18,3]])

print("Predicted Salary:", salary[0])
```

Q1. What is Multiple Linear Regression? \
Answer: A supervised learning algorithm that predicts a continuous numerical value using two or more input features.

Q2. Can Multiple Linear Regression use only one feature? \
Answer: Yes, but then it becomes simple Linear Regression.

Q3. What happens if you provide fewer features during prediction? \
Answer: Scikit-learn raises a ValueError because the number of input features must match the number used during training.

Q4. What do model.coef_ and model.intercept_ represent? \
Answer: model.coef_ contains the learned coefficients (weights) for each feature, and model.intercept_ is the constant term in the regression equation.
