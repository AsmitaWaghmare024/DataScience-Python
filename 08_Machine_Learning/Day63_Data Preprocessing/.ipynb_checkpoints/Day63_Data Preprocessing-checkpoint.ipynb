# Day63_Data Preprocessing

## What is Data Preprocessing?
Definition:Data Preprocessing is the process of cleaning and preparing raw data before training a Machine Learning model.

**Example:** \
Suppose teacher wants to predict exam marks.
She collects this data:

| Name | Hours | Marks |
| ---- | ----: | ----: |
| Ali  |     5 |    80 |
| Sara |     ? |    90 |
| John |     6 |    85 |
| Emma |     4 |     ? |

Problems: \
Missing values \
Text (Name) \
Different data types 

Can a Machine Learning model learn properly? \
No!

First, we must clean the data. \
This cleaning process is called *Data Preprocessing*.

### Why is Data Preprocessing Needed?
Without preprocessing: \
Missing values \
Wrong values \
Duplicate rows \
Text where numbers are needed \
Different scales

Result: Poor predictions

### Data Preprocessing Workflow

Raw Data \
    │ \
    ▼ \
Clean Data \
    │ \
    ▼ \
Handle Missing Values \
    │ \
    ▼ \
Convert Text to Numbers \
    │ \
    ▼ \
Scale Features \
    │ \
    ▼ \
Ready for Machine Learning

### Step 2 – What are Features (X) and Target (y)?
This is the most important concept in Machine Learning.

Suppose our dataset is:

| Hours | Attendance | Marks |
| ----: | ---------: | ----: |
|     2 |         80 |    35 |
|     4 |         90 |    55 |
|     6 |         95 |    80 |

Question: What are we trying to predict? \
Marks

So, \
Marks = Target (y)

Everything else is used to predict marks. \
Hours and Attendance = Features (X)

Memory Trick \
X → Input \
y → Output

### Step 3 – Separate Features and Target

```python
import pandas as pd

data = {
    "Hours": [2,4,6],
    "Attendance":[80,90,95],
    "Marks":[35,55,80]
}

df = pd.DataFrame(data)

X = df[["Hours","Attendance"]]

y = df["Marks"]

print("Features (X):")
print(X)

print("\nTarget (y):")
print(y)
```

*Create DataFrame \
df = pd.DataFrame(data)* \
Creates the table.

Features \
*X = df[["Hours","Attendance"]]*
Why double brackets? \
Because we are selecting multiple columns.

Target \
*y = df["Marks"]* \
Why single brackets? \
Because we select only one column.

### Step 4 – Missing Values

## Step 5 – Fill Missing Values

Dataset

| Hours |
| ----: |
|     2 |
|     4 |
|   NaN |
|     6 |

Fill using mean.

```python
df["Hours"] = df["Hours"].fillna(df["Hours"].mean())

print(df)
```

*df["Hours"]* Select Hours column. \
*.mean()* Calculates average. \
*fillna()* Replaces NaN with the mean. 

## Step 6 – Encoding Text

Machine Learning understands numbers, not text.

| Gender |
| ------ |
| Male   |
| Female |
| Male   |

Convert: \
Male → 1 \
Female → 0

```python
import pandas as pd

data = {
    "Hours": [2, 4, None, 6],
    "Gender": ["Male", "Female", "Male", "Female"],
    "Marks": [35, 55, 60, 80]
}

df = pd.DataFrame(data)

print(df)
```

```python
df["Gender"] = df["Gender"].map({
    "Male": 1,
    "Female": 0
})

print(df)
```

.map() Replaces one value with another using a dictionary.

## Step 7 – Feature Scaling

Suppose:

| Age | Salary |
| --: | -----: |
|  20 |  50000 |
|  25 |  90000 |

Salary is much larger than Age.

Machine Learning may think Salary is more important just because the numbers are larger.

We solve this using Feature **Scaling**.

### StandardScaler

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

scaled = scaler.fit_transform(df[["Hours"]])

print(scaled)
```

scaler = StandardScaler() Creates a scaler. 

Fit and Transform \
scaled = scaler.fit_transform(df[["Hours"]]) 

This does two things: 
    
1. fit() \
Learns: \
Mean \
Standard Deviation

2. transform() \
Converts values into a similar scale.

### Why Scale Data?

Without scaling: \
Age = 20 \
Salary = 50000

After scaling: \
Age = -0.75 \
Salary = 0.84 \
Now both features have similar importance.

## Mini Project ( Complete Preprocessing)

```python
import pandas as pd

data = {
    "Hours":[2,4,None,6],
    "Gender":["Male","Female","Male","Female"],
    "Marks":[35,55,60,80]
}

df = pd.DataFrame(data)

print("Original Dataset")
print(df)

df["Hours"] = df["Hours"].fillna(df["Hours"].mean())

df["Gender"] = df["Gender"].map({
    "Male":1,
    "Female":0
})

X = df[["Hours","Gender"]]
y = df["Marks"]

print("\nProcessed Features")
print(X)

print("\nTarget")
print(y)
```

Q1. What is Data Preprocessing? \
Answer: It is the process of cleaning and preparing data before training a Machine Learning model.

Q2. What are Features and Target? \
Answer: \
Features (X): Input variables used for prediction. \
Target (y): Output variable that the model predicts. 

Q3. Why can't Machine Learning models use text directly? \
Answer: Most ML algorithms work with numerical data, so text must be converted into numbers using encoding techniques.

Q4. Why is Feature Scaling important? \
Answer: It brings features to a similar scale so that variables with larger values don't dominate the learning process.
