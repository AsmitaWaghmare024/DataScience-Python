# Decision Trees

Decision Trees make decisions by asking simple questions, just like a game of Questions (Akinator).

## What is a Decision Tree?

Definition: A Decision Tree is a supervised Machine Learning algorithm that predicts an answer by asking a series of questions.

It can be used for: \
✅ Classification \ 
✅ Regression 

Today we'll focus on Classification.

**Real-Life Examples** \
Decision Trees are used for: \
Loan Approval \
Disease Prediction \
Email Spam Detection \
Student Pass/Fail \
Customer Purchase Prediction 

### Decision Tree Structure
Every Decision Tree has three main parts.

1. Root Node

The first question. \
Example: \
Study Hours > 5?

2. Internal Node

More questions. \
Attendance > 80? \

3. Leaf Node

Final answer. \
Pass or Fail 

### Our Dataset


```python
import pandas as pd

data = {
    "Hours":[2,3,4,6,7,8],
    "Attendance":[60,70,75,85,90,95],
    "Pass":[0,0,0,1,1,1]
}

df = pd.DataFrame(data)

print(df)
```

**Explanation** \
DecisionTreeClassifier \

## Step 2: Separate Features and Target

```python
X = df[["Hours","Attendance"]]

y = df["Pass"]
```

**Features** \
Hours \
Attendance 

**Target** \
Pass

## Step 3: Split Dataset

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=42
)
```

Nothing new here. \

## Step 4: Create Model

```python
model = DecisionTreeClassifier()
```

This creates an empty Decision Tree. \

## Step 5: Train Model

```python
model.fit(X_train, y_train)
```

The model starts asking: \
Which question best separates Pass and Fail?

Example: \

If yes: \
Likely Pass

If no: \
Likely Fail
    
## Step 6: Predict

Suppose a student has: \
Hours = 7 \
Attendance = 88

```python
prediction = model.predict([[7,88]])

print(prediction)
```

Meaning: Pass 

```python
import pandas as pd

from sklearn.tree import DecisionTreeClassifier

from sklearn.model_selection import train_test_split

data = {
    "Hours":[2,3,4,6,7,8],
    "Attendance":[60,70,75,85,90,95],
    "Pass":[0,0,0,1,1,1]
}

df = pd.DataFrame(data)

X = df[["Hours","Attendance"]]
y = df["Pass"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=42
)

model = DecisionTreeClassifier()

model.fit(X_train,y_train)

prediction = model.predict([[7,88]])

print("Prediction:", prediction)
```

## What is Gini Impurity?

Decision Trees need to decide: Which question should I ask first?
One method is **Gini Impurity**.

Imagine a box:\
Pass \
Pass \
Pass \
Fail 

This box is mixed.

Gini tells us how mixed the classes are. \
Lower Gini = Better split \
Higher Gini = More mixed \
Scikit-learn uses Gini by default.

## What is Entropy?
Entropy also measures disorder.

If all students pass: \
Pass \
Pass \
Pass \
Pass
Entropy = 0 \
Everything is pure.

If half pass and half fail: \
Pass \
Fail \
Pass \
Fail \
Entropy is higher because there is more uncertainty

### Gini vs Entropy

| Gini                    | Entropy              |
| ----------------------- | -------------------- |
| Default in scikit-learn | Optional             |
| Faster                  | Slightly slower      |
| Measures impurity       | Measures uncertainty |

Using Entropy:

```python
model = DecisionTreeClassifier(
    criterion="entropy"
)
```

Without specifying anything:

```python
model = DecisionTreeClassifier()
```

It uses: criterion="gini"

## What is Information Gain?

Suppose we ask: Hours > 5? \
Before asking: Mixed students. \
After asking: Two cleaner groups.

The improvement is called **Information Gain**.

Higher Information Gain = Better question.

## Mini Project (Loan Approval)

```python
import pandas as pd

from sklearn.tree import DecisionTreeClassifier

data = {
    "Income":[20000,30000,50000,70000,90000],
    "Credit":[500,600,700,750,800],
    "Approved":[0,0,1,1,1]
}

df = pd.DataFrame(data)

X = df[["Income","Credit"]]

y = df["Approved"]

model = DecisionTreeClassifier()

model.fit(X,y)

income = float(input("Enter Income: "))
credit = float(input("Enter Credit Score: "))

prediction = model.predict([[income,credit]])

if prediction[0]==1:
    print("Loan Approved")
else:
    print("Loan Rejected")
```
